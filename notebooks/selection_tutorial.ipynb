## Selecting rows using `mast-table` API

In this tutorial, we will become familiar with the different API calls for selecting rows in `mast-table`.

Let's use `astroquery.mast.MastMissions` to search for JWST observations of M4:

```python
import astropy.units as u
from astropy.coordinates import SkyCoord

from astroquery.mast import MastMissions
from mast_table import MastTable

# querying for JWST observations of M4 within 0.1 degree
mission = 'jwst'
target = SkyCoord.from_name("M4")
mast = MastMissions(mission=mission)
obs = mast.query_region(target, radius=0.1*u.deg)
```

Now, we will open up the data in a `mast-table` displayed in a sidecar splitting the bottom of the screen:

```python
mt = MastTable(obs, mission=mission)
mt.show("sidecar:split-bottom")
```

### Selection with `.select()`

We can use the `.select()` method on our `mast-table` to select based on
* index values (int, list, slice)
* table masks (Boolean arrays)
* row values for a given column (value AND column name)

We can also remove any existing selections by calling `.select()` with no arguments passed in.

```python
# single index value
mt.select(5)
```

```python
# index value list
mt.select([2,7,8])
```

```python
# slice of indices
mt.select(slice(3,7))
```

```python
# Boolean table mask
mt.select(
    mt.table["duration"] > 515.365
)
```

```python
# row values for given column
mt.select("NIRISS", column = "instrume")
```

```python
# reset selections by passing in no arg
mt.select()
```

### Selection with `.selected_mask`

We can also directly set a Boolean mask to use for selection by setting the `mast-table` `.selected_mask` attribute:

```python
# setting `selected mask` attr
mt.selected_mask = mt.table["targ_dec"] < -26.5
```

### Selection with `.selected_rows`

We can set selections based on row indices by setting the `mast-table` `.selected_rows` attribute with:
- lists of row indices
- slices of table items

```python
# set selected_rows directly
mt.selected_rows = [0,1]
```

```python
# set selected_rows with slices
mt.selected_rows = mt.items[3:5]
```
